```python
import numpy as np
import matplotlib.pyplot as plt

logit_data = {
    "spoon": 1,
    "fork": 0,
    "pot": -4,
    "cup": -2,
}

x = list(logit_data.keys())
logits = np.array(list(logit_data.values()))

def softmax(logits, T=1):
    probas = np.exp(logits / T)
    return probas / np.sum(probas + 1e-3)

probas = softmax(logits, T=1)

fig, axs = plt.subplots(1, 2, layout='constrained', figsize=(6,2))
axs[0].set_title('Logits')
axs[0].plot(x, logits, linestyle='-', color='tab:green')
axs[0].grid(True)
axs[0].set_xlabel('class',  fontsize=10)
axs[0].set_ylabel('logit',  fontsize=10)

axs[1].set_title('Probas')
axs[1].plot(x, probas, linestyle='-', color='tab:blue')
axs[1].grid(True)
axs[1].set_xlabel('class',  fontsize=10)
axs[1].set_ylabel('proba',  fontsize=10)
pass
```
