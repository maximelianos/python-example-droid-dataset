```python
"""
We'll use regular numpy to set pixel values
and draw the probability of X and Y on the edge of the image.
"""

import numpy as np
import matplotlib.pyplot as plt

height, width = 20, 20
img = np.zeros((height, width, 3), dtype=np.uint8)

probas_x = np.arange(0, width)
probas_x = probas_x / width * 255.0
probas_y = np.random.randint(0, height, height)
probas_y = probas_y / height * 255.0

stripe_width = 10 # in pixels
color_x = np.array([0.5, 0.5, 1.0])
img[0:stripe_width, :] = probas_x[:, None] * color_x[None, :] # stripe shape (stripe_width (broadcast), width, 3)

color_y = np.array([0.5, 0.5, 0.0])
img[:, 0:stripe_width] = (probas_y[:, None] * color_y[None, :])[:height, None, :3] # stripe shape (height, stripe_width (broadcast), 3)
```

```python
fig, ax = plt.subplots(1, figsize=(8,3))
ax.set_axis_off()
ax.imshow(img)
```
